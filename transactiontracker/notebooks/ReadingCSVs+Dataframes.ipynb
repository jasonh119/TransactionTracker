```python
import pandas as pd
path = "D:\\01_Data\\OneDrive\\Development\\Python\\TestData\\"
file1 = 'CutDown_TransTestData.csv'
file2 = 'RawData_SG_DailyAcc_Dec.csv'
file_path = (path + file2)
#chuck_size = 10
#print(filepath)

with open(file_path, "r") as file:
    lines = file.readlines()

account_number = lines[3].split(",")[1].strip()
account_number = account_number[1:] # Remove the first character
print("Extracted Account Number:", account_number)

account_name = lines[3].split(",")[0].strip()
print("Extract Acc Name in VS:", account_name)
```

```python
df = pd.read_csv(file_path, header=3)  # Load the entire file

df["Account"] = account_number
df["Account Name"] = account_name
df["Financial Institution"] = 'SCB'

print("Columns: ***", df.columns)
print('Columns End...')
print(df.head())  # Display the first few rows
```
